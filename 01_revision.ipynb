```python
import pandas as pd
from pathlib import Path

carpeta = Path("datos")

def leer(ruta):
    for enc in ("utf-8", "latin-1"):
        try:
            return pd.read_csv(ruta, encoding=enc, low_memory=False), enc
        except UnicodeDecodeError:
            continue
    raise ValueError(f"No se pudo leer {ruta}")

archivos = sorted(carpeta.rglob("*.csv"))
print(f"Archivos encontrados: {len(archivos)}")

for ruta in archivos:
    df, enc = leer(ruta)
    print(f"\n=== {ruta.relative_to(carpeta)} | {enc} | {len(df):,} filas")
    print("Columnas:", list(df.columns))
    print("Nulos:", df.isna().sum().to_dict())
    if "Genero" in df.columns:
        print("Genero:", df["Genero"].astype(str).value_counts().head(5).to_dict())
    print(df.head(2).to_string())
```
